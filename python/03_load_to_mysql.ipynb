```python
import pandas as pd
from sqlalchemy import create_engine

# ---- 1. Load the saved star schema CSVs ----
dim_date = pd.read_csv('dim_date.csv')
dim_customers = pd.read_csv('dim_customers.csv')
dim_products = pd.read_csv('dim_products.csv')
dim_sellers = pd.read_csv('dim_sellers.csv')
dim_payment = pd.read_csv('dim_payment.csv')
dim_order_status = pd.read_csv('dim_order_status.csv')
fact_order_items = pd.read_csv('fact_order_items.csv')

# ---- 2. Connect to MySQL ----
username = 'root'
password = 'admin'   # replace with your real password
host = 'localhost'
port = '3306'
database = 'olist'   # create this in Workbench first: CREATE DATABASE olist;

engine = create_engine(f'mysql+pymysql://{username}:{password}@{host}:{port}/{database}')

# ---- 3. Push tables to MySQL ----
dim_date.to_sql('dim_date', engine, if_exists='replace', index=False)
dim_customers.to_sql('dim_customers', engine, if_exists='replace', index=False)
dim_products.to_sql('dim_products', engine, if_exists='replace', index=False)
dim_sellers.to_sql('dim_sellers', engine, if_exists='replace', index=False)
dim_payment.to_sql('dim_payment', engine, if_exists='replace', index=False)
dim_order_status.to_sql('dim_order_status', engine, if_exists='replace', index=False)
fact_order_items.to_sql('fact_order_items', engine, if_exists='replace', index=False)

print("All tables loaded into MySQL database 'olist'")
```
